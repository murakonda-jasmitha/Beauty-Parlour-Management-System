```python
def add_customer():

    print("\n" + "="*40)
    print("          ADD CUSTOMER")
    print("="*40)

    # Customer Name
    while True:

        name = input("Enter Customer Name : ").strip()

        if name.replace(" ", "").isalpha():
            break

        else:
            print(" Name should contain only alphabets.")

    # Phone Number
    while True:

        phone = input("Enter Phone Number : ")

        if len(phone) == 10 and phone.isdigit() and phone[0] in "6789":

            # Duplicate Phone Check
            cursor.execute("SELECT * FROM customers WHERE phone=%s", (phone,))

            if cursor.fetchone():
                print(" Phone Number Already Exists.")
            else:
                break

        else:
            print(" Invalid Phone Number.")
            print("Phone number must contain 10 digits and start with 6,7,8 or 9.")

    # Age
    while True:

        try:

            age = int(input("Enter Age : "))

            if 1 <= age <= 100:
                break

            else:
                print(" Enter age between 1 and 100.")

        except:
            print(" Age should be a number.")

    # Gender
    while True:

        gender = input("Enter Gender (Male/Female/Other): ").capitalize()

        if gender in ["Male", "Female", "Other"]:
            break

        else:
            print(" Invalid Gender.")

    # Address
    address = input("Enter Address : ")

    try:

        query = """
        INSERT INTO customers
        (customer_name, phone, age, gender, address)
        VALUES(%s,%s,%s,%s,%s)
        """

        values = (name, phone, age, gender, address)

        cursor.execute(query, values)

        conn.commit()

        print("\n Customer Added Successfully.")

    except Exception as e:

        print("Error :", e)
```

```python
def generate_bill():

    print("\n" + "="*45)
    print("          GENERATE BILL")
    print("="*45)

    appointment_id = int(input("Enter Appointment ID : "))

    # Check appointment exists and status
    cursor.execute("""
    SELECT status
    FROM appointments
    WHERE appointment_id=%s
    """, (appointment_id,))

    result = cursor.fetchone()

    if result is None:
        print("\n Appointment Not Found.")
        return

    if result[0] != "Completed":
        print("\n Complete the appointment before generating the bill.")
        return

    # Check bill already generated
    cursor.execute("""
    SELECT *
    FROM bills
    WHERE appointment_id=%s
    """, (appointment_id,))

    if cursor.fetchone():
        print("\n❌ Bill Already Generated.")
        return

    # Fetch bill details
    query = """
    SELECT
        c.customer_name,
        e.employee_name,
        s.service_name,
        s.price

    FROM appointments a

    JOIN customers c
    ON a.customer_id = c.customer_id

    JOIN employees e
    ON a.employee_id = e.employee_id

    JOIN services s
    ON a.service_id = s.service_id

    WHERE a.appointment_id=%s
    """

    cursor.execute(query, (appointment_id,))

    bill = cursor.fetchone()

    amount = float(bill[3])
    gst = amount * 0.18
    total = amount + gst

    bill_date = datetime.now().date()

    cursor.execute("""
    INSERT INTO bills
    (appointment_id, bill_date, total_amount)
    VALUES(%s,%s,%s)
    """, (appointment_id, bill_date, total))

    conn.commit()

    print("\n" + "="*50)
    print("           BEAUTY PARLOUR BILL")
    print("="*50)

    print(f"Customer Name : {bill[0]}")
    print(f"Employee Name : {bill[1]}")
    print(f"Service       : {bill[2]}")
    print(f"Amount        : ₹{amount:.2f}")
    print(f"GST (18%)     : ₹{gst:.2f}")
    print("-"*50)
    print(f"Total Amount  : ₹{total:.2f}")
    print(f"Payment Mode  : Cash")
    print(f"Payment Status: Paid")
    print(f"Bill Date     : {bill_date}")

    print("="*50)
    print("      Thank You... Visit Again!")
    print("="*50)
```

```python
from datetime import date

def make_payment():

    print("\n" + "="*40)
    print("          MAKE PAYMENT")
    print("="*40)

    bill_id = int(input("Enter Bill ID : "))

    cursor.execute(
        "SELECT * FROM bills WHERE bill_id=%s",
        (bill_id,)
    )

    bill = cursor.fetchone()

    if bill:

        payment_mode = "Cash"
        payment_status = "Paid"
        payment_date = date.today()

        cursor.execute("""
        INSERT INTO payments
        (bill_id, payment_mode, payment_status, payment_date)
        VALUES(%s,%s,%s,%s)
        """,
        (bill_id, payment_mode, payment_status, payment_date))

        conn.commit()

        print("\n Payment Successful")
        print("Payment Mode :", payment_mode)
        print("Payment Status :", payment_status)

    else:

        print("\n Bill Not Found.")
```

```python
def payment_history():

    print("\n" + "="*90)
    print("                 PAYMENT HISTORY")
    print("="*90)

    query = """
    SELECT
        p.payment_id,
        p.bill_id,
        p.payment_mode,
        p.payment_status,
        p.payment_date

    FROM payments p
    ORDER BY p.payment_id
    """

    cursor.execute(query)

    payments = cursor.fetchall()

    if not payments:

        print("\nNo Payment History Found.")
        return

    print(f"{'Payment ID':<12}{'Bill ID':<10}{'Mode':<15}{'Status':<15}{'Date'}")
    print("-"*90)

    for payment in payments:

        print(f"{payment[0]:<12}{payment[1]:<10}{payment[2]:<15}{payment[3]:<15}{payment[4]}")
```

```python
from datetime import date

def today_income():

    print("\n" + "="*40)
    print("         TODAY'S INCOME")
    print("="*40)

    today = date.today()

    query = """
    SELECT IFNULL(SUM(total_amount),0)
    FROM bills
    WHERE bill_date=%s
    """

    cursor.execute(query, (today,))

    income = cursor.fetchone()[0]

    print(f"\nToday's Income : ₹{income:.2f}")
```

```python
def monthly_income():

    print("\n" + "="*40)
    print("        MONTHLY INCOME")
    print("="*40)

    month = input("Enter Month (MM) : ")
    year = input("Enter Year (YYYY) : ")

    query = """
    SELECT IFNULL(SUM(total_amount),0)
    FROM bills
    WHERE MONTH(bill_date)=%s
    AND YEAR(bill_date)=%s
    """

    cursor.execute(query, (month, year))

    income = cursor.fetchone()[0]

    print(f"\nMonthly Income : ₹{income:.2f}")
```

```python
def most_popular_service():

    print("\n" + "="*45)
    print("      MOST POPULAR SERVICE")
    print("="*45)

    query = """
    SELECT
        s.service_name,
        COUNT(*) AS total_bookings

    FROM appointments a

    JOIN services s
    ON a.service_id = s.service_id

    GROUP BY s.service_name

    ORDER BY total_bookings DESC

    LIMIT 1
    """

    cursor.execute(query)

    result = cursor.fetchone()

    if result:

        print("Service Name :", result[0])
        print("Bookings     :", result[1])

    else:

        print("No Data Found.")
```
